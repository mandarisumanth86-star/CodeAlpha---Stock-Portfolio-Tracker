```python
# CodeAlpha Task 2 - Stock Portfolio Tracker

stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 170,
    "MSFT": 420,
    "AMZN": 150
}

portfolio = {}
total_investment = 0

print("=" * 40)
print("     STOCK PORTFOLIO TRACKER")
print("=" * 40)

while True:
    stock = input("\nEnter Stock Symbol (AAPL/TSLA/GOOGL/MSFT/AMZN): ").upper()

    if stock not in stock_prices:
        print("Stock not available.")
        continue

    quantity = int(input("Enter Quantity: "))

    portfolio[stock] = portfolio.get(stock, 0) + quantity

    choice = input("Do you want to add another stock? (yes/no): ").lower()
    if choice != "yes":
        break

print("\n" + "=" * 40)
print("PORTFOLIO SUMMARY")
print("=" * 40)

report = []
report.append("STOCK PORTFOLIO SUMMARY\n")
report.append("-" * 40 + "\n")

for stock, quantity in portfolio.items():
    price = stock_prices[stock]
    value = price * quantity
    total_investment += value

    print(f"{stock} : {quantity} shares × ${price} = ${value}")

    report.append(f"{stock}: {quantity} shares × ${price} = ${value}\n")

print("-" * 40)
print(f"Total Investment = ${total_investment}")

report.append("-" * 40 + "\n")
report.append(f"Total Investment = ${total_investment}")

with open("portfolio_summary.txt", "w") as file:
    file.writelines(report)

print("\nPortfolio saved successfully in portfolio_summary.txt")
```
